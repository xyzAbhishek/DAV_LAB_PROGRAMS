```python
import numpy as np
import pandas as pd
```

```python
arr=np.array([25000,32000,28000,35000,38000])
print(arr)
```

```python
print("Total Sales:",np.sum(arr))
print("Average Sales:",np.mean(arr))
print("Maximum Sales:",np.max(arr))
print("Minimum Sales:",np.min(arr))
print("Display sales from third month:",arr[2:])
print("Display every second month's Sales:",arr[1::2])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Student_Marks.csv")
```

```python
print(df)
```

```python
df.hist(figsize=(10, 6), bins=10, edgecolor="brown")

plt.suptitle("Histograms of Student Data", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
sns.heatmap(df[["Marks"]], cmap="YlGnBu")

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df)

plt.show()
```

```python
plt.scatter(df["time_study"], df["Marks"])

plt.xlabel("Time Study")
plt.ylabel("Marks")
plt.title("Study Time vs Marks")
plt.show()
```
